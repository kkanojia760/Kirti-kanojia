# **Question 1: What is the role of filters and feature maps in Convolutional Neural Network (CNN)?**


## Role of Filters and Feature Maps in Convolutional Neural Networks (CNNs)

In Convolutional Neural Networks (CNNs), filters (also known as kernels) and feature maps are fundamental components that enable the network to learn and detect hierarchical patterns in input data, such as images.

### Filters (Kernels)

1.  **Detection of Features**: A filter is a small matrix of numbers that slides (convolves) over the input data (e.g., an image). Each filter is designed or learned to detect a specific type of feature or pattern, such as edges, corners, textures, or more complex shapes.

2.  **Parameter Sharing**: The same filter is applied across the entire input. This property, known as parameter sharing, significantly reduces the number of parameters in the network, making CNNs more efficient and less prone to overfitting compared to traditional fully connected networks.

3.  **Spatial Invariance**: By applying the same filter across different locations, CNNs achieve a degree of spatial invariance. This means that if a particular feature appears in a different part of the image, the network can still detect it.

4.  **Learning**: During the training process, the values within these filter matrices are learned through backpropagation and gradient descent. The network adjusts these values to optimize its ability to detect relevant features for the given task (e.g., image classification).

### Feature Maps (Activation Maps)

1.  **Output of Convolution**: When a filter is convolved with the input data, it produces an output called a feature map (or activation map). This feature map shows where the specific feature that the filter is looking for is present in the input, and with what intensity.

2.  **Encoding Spatial Information**: Each pixel in a feature map corresponds to a specific region in the input image. The value of this pixel indicates the strength of activation of the feature detected by the filter in that region.

3.  **Hierarchical Representation**: As the data passes through successive convolutional layers, the feature maps become increasingly abstract. Early layers might detect simple features like edges, while deeper layers combine these simple features to detect more complex patterns, such as parts of objects (e.g., an eye, a wheel) and eventually entire objects.

4.  **Input for Next Layer**: The feature maps generated by one convolutional layer serve as the input for the subsequent layers (either another convolutional layer or a pooling layer).

### Analogy

Imagine you're trying to find specific shapes in a painting. Your eyes are like filters, each trained to spot a particular shape (e.g., a circle, a square). As your eyes scan the painting, you create a mental 'map' for each shape, indicating where and how strongly that shape appears. These mental maps are like feature maps, and your trained eyes are the filters.

# **Question 2: Explain the concepts of padding and stride in CNNs(Convolutional Neural  Network). How do they affect the output dimensions of feature maps?**

## Padding and Stride in Convolutional Neural Networks (CNNs)

Padding and stride are two crucial concepts in Convolutional Neural Networks that control the size of the output feature map and how the convolution operation is performed.

### Padding

Padding refers to adding extra pixels (typically zeros) around the border of the input image or feature map before applying the convolution operation. This is done for a few key reasons:

1.  **Preserving Spatial Dimensions**: Without padding, applying a convolution filter often reduces the spatial dimensions (height and width) of the output feature map. Padding helps to retain the original input size, preventing the feature map from shrinking too quickly in deeper layers.

2.  **Retaining Edge Information**: Pixels at the edges of an image are only involved in a few convolution operations, meaning their information might not be fully captured. Padding ensures that edge pixels are processed multiple times, giving the model more information about the boundaries and corners of the input.

3.  **Controlling Output Size**: By adjusting the amount of padding, you can precisely control the spatial dimensions of the output feature map.

There are generally two types of padding:
*   **Valid Padding (No Padding)**: No padding is added. The filter only slides over the valid positions of the input. The output feature map will be smaller than the input.
*   **Same Padding**: Padding is added such that the output feature map has the same spatial dimensions as the input feature map (assuming a stride of 1).

### Stride

Stride defines the number of pixels the filter shifts over the input image or feature map at each step during the convolution operation. It controls the density of information extraction and the size of the output feature map:

1.  **Downsampling**: A stride greater than 1 effectively downsamples the input. For example, a stride of 2 means the filter moves two pixels at a time, resulting in a smaller output feature map.

2.  **Computational Efficiency**: Larger strides reduce the number of convolution operations performed, thereby decreasing computational cost and speeding up the network.

3.  **Feature Map Size**: Stride has a direct impact on the dimensions of the output. Larger strides lead to smaller output feature maps.

### How They Affect Output Dimensions

The output spatial dimensions (height $H_{out}$ and width $W_{out}$) of a feature map after a convolutional layer can be calculated using the following formulas:

$H_{out} = \lfloor \frac{H_{in} - F + 2P}{S} \rfloor + 1$

$W_{out} = \lfloor \frac{W_{in} - F + 2P}{S} \rfloor + 1$

Where:
*   $H_{in}$, $W_{in}$: Height and width of the input feature map.
*   $F$: Size of the convolutional filter (kernel) (e.g., $3$ for a $3 \times 3$ filter).
*   $P$: Amount of padding applied (e.g., $1$ for $1$ pixel of padding on each side).
*   $S$: Stride (number of pixels the filter shifts).
*   $\lfloor \dots \rfloor$: The floor operator, which rounds down to the nearest integer.


**Example**:
Consider an input image of size $7 \times 7$, a filter of size $3 \times 3$.

*   **No Padding, Stride 1**:
    $H_{out} = \lfloor \frac{7 - 3 + 2 \times 0}{1} \rfloor + 1 = 4 + 1 = 5$
    Output size: $5 \times 5$

*   **Padding 1, Stride 1 (Same Padding)**:
    $H_{out} = \lfloor \frac{7 - 3 + 2 \times 1}{1} \rfloor + 1 = 6 + 1 = 7$
    Output size: $7 \times 7$

*   **No Padding, Stride 2**:
    $H_{out} = \lfloor \frac{7 - 3 + 2 \times 0}{2} \rfloor + 1 = \lfloor 2 \rfloor + 1 = 3$
    Output size: $3 \times 3$

Understanding padding and stride is essential for designing effective CNN architectures, as they directly control the flow of spatial information and the overall depth and breadth of the network.

# **Question 3: Define receptive field in the context of CNNs. Why is it important for deep architectures?**

## Receptive Field in Convolutional Neural Networks (CNNs)

In the context of Convolutional Neural Networks, the **receptive field** refers to the region in the input space (e.g., the input image) that a particular neuron in a feature map is looking at. More precisely, it's the area of the input image that influences the output of a specific neuron in a convolutional layer. Every neuron in a feature map only 'sees' a localized region of the input through its filter.

### How the Receptive Field is Formed

The receptive field of a neuron is determined by:

1.  **Filter Size (Kernel Size)**: A larger filter size means a larger immediate receptive field for neurons in that layer.
2.  **Stride**: A larger stride causes the receptive fields of subsequent neurons in the feature map to overlap less and spread out faster across the input.
3.  **Padding**: Padding doesn't directly increase the *size* of the receptive field for a given neuron, but it affects how much of the input's edges are included in the computations.
4.  **Number of Layers**: As you go deeper into the network, the receptive field of neurons in later layers expands significantly. A neuron in a deep layer is influenced by a much larger area of the original input image because it processes the combined outputs of many neurons from previous layers, each with their own receptive fields.

### Importance for Deep Architectures

The concept of the receptive field is crucial for deep architectures for several reasons:

1.  **Hierarchical Feature Learning**: Deep CNNs are designed to learn hierarchical representations. Early layers have small receptive fields, allowing them to detect fine-grained, local features like edges and corners. As the network deepens, neurons in later layers have progressively larger receptive fields, enabling them to integrate these local features into more complex and abstract patterns, such as object parts or even entire objects.

2.  **Contextual Information**: A larger receptive field allows a neuron to gather more contextual information from the input. For example, to recognize a 'face,' a neuron needs to see not just an 'eye' or a 'nose' but their spatial arrangement relative to each other, which requires a larger receptive field that covers a significant portion of the face.

3.  **Understanding Global Patterns**: In tasks like image classification or object detection, understanding global patterns and relationships between different parts of an image is vital. The expanding receptive fields in deep layers enable the network to capture these global patterns, leading to more robust and accurate predictions.

4.  **Efficiency in Information Processing**: Instead of using a single large filter to cover the entire image (which would be computationally expensive and suffer from a high number of parameters), deep architectures stack multiple smaller filters. This strategy is more efficient, as two $3 \times 3$ convolutional layers (with stride 1 and 'same' padding) produce a receptive field equivalent to a $5 \times 5$ filter, but with fewer parameters and non-linearity introduced by activation functions between layers.

5.  **Dealing with Varying Scales**: The hierarchical nature of receptive fields allows CNNs to handle objects that appear at various scales in an image. Features detected at smaller scales by shallower layers can be combined and recognized at larger scales by deeper layers.

In essence, the receptive field explains how a CNN neuron "sees" and processes information from its input. The progressive expansion of the receptive field through deep layers is fundamental to the ability of CNNs to learn complex, high-level features and achieve remarkable performance in various computer vision tasks.

# **Question 4: Discuss how filter size and stride influence the number of parameters in a CNN.**


## How Filter Size and Stride Influence the Number of Parameters in a CNN

In a Convolutional Neural Network (CNN), the number of parameters is a critical factor influencing model complexity, memory footprint, and computational cost. Filter size and stride directly affect this count in different ways.

### Filter Size (Kernel Size)

1.  **Direct Impact on Parameters per Filter**: Each filter (kernel) is a small matrix of learnable weights. The number of parameters *within a single filter* is directly proportional to its size. For a filter of size $F \times F$ operating on an input with $C_{in}$ channels, and producing $C_{out}$ output channels (i.e., using $C_{out}$ distinct filters), the number of parameters introduced by these filters in one convolutional layer is:
    
    $P_{\text{filters}} = (F \times F \times C_{in} + 1) \times C_{out}$
    
    The `+ 1` accounts for the bias term associated with each filter. If there are no biases, it's simply $F \times F \times C_{in} \times C_{out}$.

    *   **Larger Filters = More Parameters**: A larger filter size ($F$) significantly increases the number of parameters. For example, a $5 \times 5$ filter has 25 weights, while a $3 \times 3$ filter has 9 weights. If you use many such filters, the parameter count grows quadratically with $F$.

2.  **Trade-off with Depth**: While larger filters capture broader spatial information in a single step, modern CNN architectures often prefer stacking multiple smaller filters (e.g., two $3 \times 3$ filters instead of one $5 \times 5$ filter). This approach achieves a similar or even larger receptive field with fewer parameters and introduces more non-linearity, leading to more powerful feature representations.

    *   Example: For an input with 64 channels and outputting 64 channels:
        *   One $5 \times 5$ filter: $(5 \times 5 \times 64 + 1) \times 64 \approx 102,464$ parameters.
        *   Two $3 \times 3$ filters: $2 \times ((3 \times 3 \times 64 + 1) \times 64) \approx 73,792$ parameters. (Fewer parameters for a similar receptive field).

### Stride

1.  **No Direct Impact on Parameter Count**: Stride, which determines how many pixels the filter moves at each step, **does not directly influence the number of learnable parameters** (weights and biases) within the convolutional filters themselves. The filters retain their size and internal weights regardless of the stride.

2.  **Indirect Impact via Feature Map Size**: While stride doesn't change the parameters, it significantly affects the spatial dimensions of the output feature map. A larger stride reduces the size of the feature map. This, in turn, can indirectly affect the number of parameters in **subsequent layers** (especially fully connected layers that often follow convolutional layers).

    *   **Reduced Feature Map Size -> Fewer Parameters in Subsequent Layers**: If a larger stride leads to a smaller feature map, and this feature map is then flattened and fed into a fully connected layer, the number of input neurons to the fully connected layer (and thus its parameters) will be substantially smaller. This is a common technique for downsampling while reducing computational load and parameter count in the latter parts of the network.

    *   Example: If a $10 \times 10$ feature map is flattened, it becomes 100 values. If a stride reduces it to a $5 \times 5$ feature map, it becomes 25 values, leading to significantly fewer connections and parameters in the next fully connected layer.

### Summary

*   **Filter Size**: Directly and quadratically impacts the number of parameters *within* the convolutional layer itself. Larger filters mean more parameters. Strategically using smaller filters can lead to parameter efficiency.
*   **Stride**: Does **not** directly affect the number of parameters in the convolutional layer where it's applied. However, by reducing the spatial dimensions of the output feature map, it can *indirectly* reduce the number of parameters in subsequent layers, particularly in fully connected layers, thereby making the overall model more efficient.

# **Question 5: Compare and contrast different CNN-based architectures like LeNet, AlexNet, and VGG in terms of depth, filter sizes, and performance.**


## Comparison of LeNet, AlexNet, and VGG Architectures

LeNet, AlexNet, and VGG are landmark CNN architectures that have progressively pushed the boundaries of image recognition. They represent key advancements in depth, computational power, and architectural design.

### 1. LeNet-5 (1998)

*   **Pioneering Work**: One of the earliest successful CNNs, designed by Yann LeCun for handwritten digit recognition.
*   **Depth**: Relatively shallow, typically consisting of 7 layers (3 convolutional, 2 pooling, 2 fully connected).
*   **Filter Sizes**: Primarily used $5 \times 5$ filters in its convolutional layers.
*   **Key Features**:
    *   Alternating convolutional and subsampling (pooling) layers.
    *   Sigmoid/tanh activation functions.
    *   Fully connected layers at the end.
    *   Emphasized local receptive fields, shared weights, and spatial or temporal subsampling.
*   **Performance**: Achieved excellent results on the MNIST dataset but was limited by available computational resources and dataset size for more complex tasks.
*   **Era**: Pre-deep learning boom.

### 2. AlexNet (2012)

*   **Breakthrough Model**: Won the ImageNet Large Scale Visual Recognition Challenge (ILSVRC) in 2012 by a significant margin, marking the beginning of the deep learning era.
*   **Depth**: Deeper than LeNet, typically 8 layers (5 convolutional, 3 fully connected).
*   **Filter Sizes**:
    *   Started with a large $11 \times 11$ filter in the first convolutional layer (to capture large-scale features from relatively high-resolution inputs).
    *   Followed by $5 \times 5$ and then $3 \times 3$ filters in subsequent layers.
*   **Key Features**:
    *   **ReLU Activation**: Introduced Rectified Linear Units (ReLU) instead of sigmoid/tanh, which helped mitigate the vanishing gradient problem and sped up training.
    *   **Dropout**: Used dropout in fully connected layers to prevent overfitting.
    *   **Data Augmentation**: Extensive data augmentation techniques to increase dataset size and improve generalization.
    *   **GPU Implementation**: Designed to be trained on two GPUs, demonstrating the importance of parallel computing for deep models.
*   **Performance**: Achieved a top-5 error rate of 15.3% on ImageNet, a massive improvement over previous methods. Demonstrated the power of deeper networks, GPUs, and ReLU for complex image classification tasks.
*   **Era**: Early deep learning boom.

### 3. VGG (Visual Geometry Group) (2014)

*   **Simplicity and Depth**: A series of models (VGG16, VGG19) that further pushed the depth of CNNs, emphasizing simplicity and uniformity in architecture.
*   **Depth**: Very deep, with 16 or 19 layers (13 or 16 convolutional, 3 fully connected).
*   **Filter Sizes**: Consistently used very small $3 \times 3$ convolutional filters throughout the network, followed by $2 \times 2$ max-pooling layers.
*   **Key Features**:
    *   **Stacked Small Filters**: Demonstrated that stacking multiple small ($3 \times 3$) convolutional layers can achieve the same effective receptive field as a larger filter (e.g., two $3 \times 3$ layers have a $5 \times 5$ effective receptive field) but with fewer parameters and more non-linearity. This was a critical insight for deep learning design.
    *   **Homogeneous Architecture**: Maintained a very uniform structure with repeated blocks of convolution and pooling layers, making it easy to understand and implement.
*   **Performance**: Achieved excellent results in ILSVRC 2014, showing a top-5 error rate of 7.3%. Its architectural simplicity made it a popular baseline for subsequent research.
*   **Era**: Consolidation and refinement of deep learning.

### Comparison Table

| Feature           | LeNet-5                                  | AlexNet                                       | VGG16/19                                    |
| :---------------- | :--------------------------------------- | :-------------------------------------------- | :------------------------------------------ |
| **Year**          | 1998                                     | 2012                                          | 2014                                        |
| **Depth**         | Shallow (7 layers)                       | Medium (8 layers)                             | Very Deep (16 or 19 layers)                 |
| **Filter Sizes**  | Mostly $5 \times 5$                     | $11 \times 11$, $5 \times 5$, $3 \times 3$ | Exclusively $3 \times 3$                   |
| **Activation**    | Sigmoid/Tanh                             | ReLU                                          | ReLU                                        |
| **Key Innovation**| Pioneering CNN, shared weights           | ReLU, Dropout, GPU training                   | Stacked small filters, deep uniform architecture |
| **Parameters**    | ~60K                                     | ~60 Million                                   | ~138-144 Million                            |
| **Performance**   | Good for digit recognition               | Breakthrough on ImageNet                      | State-of-the-art on ImageNet (for its time) |


### Conclusion

These three architectures represent an evolution in CNN design. LeNet established the foundational concepts. AlexNet scaled these concepts with modern computational power and regularization techniques to achieve breakthrough performance. VGG then refined the architectural principles, demonstrating the power of deep, homogeneous networks with small filters. Each model built upon the successes of its predecessors, paving the way for even deeper and more complex architectures that followed.

# **Question 6: Using keras, build and train a simple CNN model on the MNIST dataset from scratch. Include code for module creation, compilation, training, and evaluation.**

### 1. Import Libraries

First, we need to import the necessary libraries from TensorFlow and Keras.

````python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
import matplotlib.pyplot as plt
````

### 2. Load and Preprocess the MNIST Dataset

We'll load the MNIST dataset, which consists of grayscale images of handwritten digits (0-9). The images are $28 \times 28$ pixels. We need to reshape them to include a channel dimension (since CNNs expect a 4D input: `[batch, height, width, channels]`), normalize the pixel values, and one-hot encode the labels.

````python
# Load the MNIST dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Reshape data to add a channel dimension (1 for grayscale images)
x_train = x_train.reshape((x_train.shape[0], 28, 28, 1)).astype('float32')
x_test = x_test.reshape((x_test.shape[0], 28, 28, 1)).astype('float32')

# Normalize pixel values to be between 0 and 1
x_train /= 255.0
x_test /= 255.0

# One-hot encode the labels
y_train = keras.utils.to_categorical(y_train, 10)
y_test = keras.utils.to_categorical(y_test, 10)

print(f"x_train shape: {x_train.shape}, y_train shape: {y_train.shape}")
print(f"x_test shape: {x_test.shape}, y_test shape: {y_test.shape}")
````

### 3. Build the CNN Model

We'll create a simple sequential CNN model. It will consist of:
*   A `Conv2D` layer with 32 filters, a $3 \times 3$ kernel, and ReLU activation.
*   A `MaxPooling2D` layer to reduce spatial dimensions.
*   Another `Conv2D` layer with 64 filters and ReLU activation.
*   Another `MaxPooling2D` layer.
*   A `Flatten` layer to convert the 2D feature maps into a 1D vector.
*   A `Dense` (fully connected) hidden layer with ReLU activation.
*   A final `Dense` output layer with 10 units (for 10 classes) and `softmax` activation.

````python
model = keras.Sequential([
    # First Convolutional Block
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),

    # Second Convolutional Block
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Flatten and Dense Layers
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax') # 10 classes for MNIST digits
])

model.summary()
````

### 4. Compile the Model

Before training, we need to compile the model. We'll specify:
*   **Optimizer**: `adam` (a popular choice for its efficiency).
*   **Loss function**: `categorical_crossentropy` (suitable for multi-class classification with one-hot encoded labels).
*   **Metrics**: `accuracy` to monitor performance during training.

````python
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
````

### 5. Train the Model

Now, we'll train the model using the `fit` method. We'll train for a few epochs and use a batch size. It's good practice to set `validation_data` to monitor the model's performance on unseen data during training.

````python
history = model.fit(x_train, y_train, epochs=5, batch_size=64, validation_data=(x_test, y_test))

# Plot training history (optional)
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.show()
````

### 6. Evaluate the Model

Finally, we'll evaluate the trained model's performance on the test dataset to get its final accuracy and loss.

````python
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
print(f'\nTest accuracy: {test_acc:.4f}')
print(f'Test loss: {test_loss:.4f}')
````

### 7. Make Predictions (Optional)

Let's visualize a few predictions from the test set.

````python
predictions = model.predict(x_test)

plt.figure(figsize=(10, 10))
for i in range(9):
    plt.subplot(3, 3, i + 1)
    plt.imshow(x_test[i].reshape(28, 28), cmap='gray')
    predicted_label = np.argmax(predictions[i])
    true_label = np.argmax(y_test[i])
    plt.title(f'Pred: {predicted_label} (True: {true_label})')
    plt.axis('off')
plt.show()
````

# **Question 7: Load and preprocess the CIFAR-10 dataset using Keras, and create a CNN model to classify RGB images. Show your preprocessing and architecture.**

### 1. Load and Preprocess the CIFAR-10 Dataset

The CIFAR-10 dataset consists of 32x32 RGB images of 10 classes. We need to normalize the pixel values to be between 0 and 1 and one-hot encode the labels. Since CIFAR-10 images are already 3D (height, width, channels), no reshaping like MNIST is needed.

````python
# Load the CIFAR-10 dataset
(x_train_cifar, y_train_cifar), (x_test_cifar, y_test_cifar) = keras.datasets.cifar10.load_data()

# Normalize pixel values to be between 0 and 1
x_train_cifar = x_train_cifar.astype('float32') / 255.0
x_test_cifar = x_test_cifar.astype('float32') / 255.0

# One-hot encode the labels
y_train_cifar = keras.utils.to_categorical(y_train_cifar, 10)
y_test_cifar = keras.utils.to_categorical(y_test_cifar, 10)

print(f"x_train_cifar shape: {x_train_cifar.shape}, y_train_cifar shape: {y_train_cifar.shape}")
print(f"x_test_cifar shape: {x_test_cifar.shape}, y_test_cifar shape: {y_test_cifar.shape}")
````

### 2. Build the CNN Model for CIFAR-10

We'll create a CNN model similar to the MNIST one, but adapted for the 32x32x3 input shape of CIFAR-10 images. We'll use more filters and potentially more layers to handle the increased complexity of color images.

````python
cifar_model = keras.Sequential([
    # First Convolutional Block
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3), padding='same'),
    layers.BatchNormalization(),
    layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),

    # Second Convolutional Block
    layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),

    # Third Convolutional Block
    layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.25),

    # Flatten and Dense Layers
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax') # 10 classes for CIFAR-10
])

cifar_model.summary()
````

### 3. Compile the CIFAR-10 Model

We'll compile the model with the Adam optimizer, categorical crossentropy loss, and accuracy metric.

### 4. Train the CIFAR-10 Model

Now, we'll train the `cifar_model` using the `fit` method with the CIFAR-10 training data and validate it against the CIFAR-10 test data. We'll also plot the training and validation accuracy and loss over epochs.

````python
history_cifar = cifar_model.fit(x_train_cifar, y_train_cifar, epochs=10, batch_size=64, validation_data=(x_test_cifar, y_test_cifar))

# Plot training history
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history_cifar.history['accuracy'], label='Training Accuracy')
plt.plot(history_cifar.history['val_accuracy'], label='Validation Accuracy')
plt.title('CIFAR-10 Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_cifar.history['loss'], label='Training Loss')
plt.plot(history_cifar.history['val_loss'], label='Validation Loss')
plt.title('CIFAR-10 Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.show()
````

### 5. Evaluate the CIFAR-10 Model

Finally, we'll evaluate the trained `cifar_model`'s performance on the CIFAR-10 test dataset to get its final accuracy and loss.

````python
test_loss_cifar, test_acc_cifar = cifar_model.evaluate(x_test_cifar, y_test_cifar, verbose=2)
print(f'\nCIFAR-10 Test accuracy: {test_acc_cifar:.4f}')
print(f'CIFAR-10 Test loss: {test_loss_cifar:.4f}')
````

# **Question 8: Using PyTorch, write a script to define and train a CNN on the MNIST dataset. Include model definition, data loaders, training loop, and accuracy evaluation.**


### 1. Import Libraries

First, we'll import the essential PyTorch libraries for building and training neural networks, as well as `torchvision` for dataset handling and transformations.

````python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import numpy as np
````

### 2. Define the CNN Model Architecture

We'll create a simple CNN model using `nn.Module`. This model will consist of two convolutional layers followed by max-pooling layers, and then two fully connected layers. ReLU activation will be used for hidden layers, and `log_softmax` for the output layer for classification.

````python
class SimpleCNN(nn.Module):
    def __init__(self):
        super(SimpleCNN, self).__init__()
        # First convolutional layer
        # Input channels: 1 (grayscale MNIST images)
        # Output channels: 32
        # Kernel size: 3x3
        self.conv1 = nn.Conv2d(1, 32, kernel_size=3, padding=1)
        # Second convolutional layer
        # Input channels: 32
        # Output channels: 64
        # Kernel size: 3x3
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        # Max-pooling layer
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        # Fully connected layers
        # After two conv/pool layers, the image size becomes (28/2)/2 = 7x7
        # And we have 64 output channels from conv2, so 64 * 7 * 7 features
        self.fc1 = nn.Linear(64 * 7 * 7, 128)
        self.fc2 = nn.Linear(128, 10) # 10 output classes for MNIST

    def forward(self, x):
        # -> conv1 -> relu -> pool
        x = self.pool(torch.relu(self.conv1(x)))
        # -> conv2 -> relu -> pool
        x = self.pool(torch.relu(self.conv2(x)))
        # Flatten the output for the fully connected layers
        x = x.view(-1, 64 * 7 * 7)
        # -> fc1 -> relu
        x = torch.relu(self.fc1(x))
        # -> fc2 (output layer)
        x = self.fc2(x)
        # Apply log_softmax for probability distribution
        return torch.log_softmax(x, dim=1)

# Instantiate the model
model = SimpleCNN()

# Move model to GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

print(f"Model will be trained on: {device}")
print(model)
````

### 3. Prepare Data Loaders

We'll load the MNIST dataset using `torchvision.datasets`. The images will be transformed to PyTorch tensors and normalized. Then, `DataLoader`s will be created for both training and testing, allowing for batch processing.

````python
# Define transformations for the MNIST dataset
transform = transforms.Compose([
    transforms.ToTensor(), # Convert PIL Image or numpy.ndarray to tensor
    transforms.Normalize((0.1307,), (0.3081,)) # Normalize pixel values
])

# Load the MNIST training and test datasets
train_dataset = datasets.MNIST('../data', train=True, download=True, transform=transform)
test_dataset = datasets.MNIST('../data', train=False, download=True, transform=transform)

# Define data loaders
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=1000, shuffle=False)

print(f"Number of training samples: {len(train_dataset)}")
print(f"Number of test samples: {len(test_dataset)}")
print(f"Number of training batches: {len(train_loader)}")
print(f"Number of test batches: {len(test_loader)}")
````

### 4. Define Loss Function and Optimizer

For multi-class classification, `NLLLoss` (Negative Log Likelihood Loss) is suitable when the model outputs `log_softmax`. We'll use the Adam optimizer.

````python
criterion = nn.NLLLoss() # Appropriate for log_softmax output
optimizer = optim.Adam(model.parameters(), lr=0.001)
````

### 5. Implement Training and Evaluation Functions

We'll define two functions: one for training the model for a single epoch and another for evaluating its performance on the test set.

````python
def train(model, device, train_loader, optimizer, epoch):
    model.train() # Set the model to training mode
    running_loss = 0.0
    for batch_idx, (data, target) in enumerate(train_loader):
        data, target = data.to(device), target.to(device)
        optimizer.zero_grad() # Zero the gradients
        output = model(data) # Forward pass
        loss = criterion(output, target) # Calculate loss
        loss.backward() # Backward pass (compute gradients)
        optimizer.step() # Update weights
        running_loss += loss.item()
        if batch_idx % 100 == 0: # Print progress every 100 batches
            print(f'Train Epoch: {epoch} [{batch_idx * len(data)}/{len(train_loader.dataset)} ({100. * batch_idx / len(train_loader):.0f}%)]\tLoss: {loss.item():.6f}')
    avg_loss = running_loss / len(train_loader)
    print(f'====> Epoch: {epoch} Average training loss: {avg_loss:.4f}')
    return avg_loss

def test(model, device, test_loader):
    model.eval() # Set the model to evaluation mode
    test_loss = 0
    correct = 0
    with torch.no_grad(): # Disable gradient calculation during evaluation
        for data, target in test_loader:
            data, target = data.to(device), target.to(device)
            output = model(data)
            test_loss += criterion(output, target).item() # Sum up batch loss
            pred = output.argmax(dim=1, keepdim=True) # Get the index of the max log-probability
            correct += pred.eq(target.view_as(pred)).sum().item()

    test_loss /= len(test_loader.dataset) # Average loss per sample

    print(f'\nTest set: Average loss: {test_loss:.4f}, Accuracy: {correct}/{len(test_loader.dataset)} ({100. * correct / len(test_loader.dataset):.0f}%)\n')
    return test_loss, 100. * correct / len(test_loader.dataset)
````

### 6. Training Loop and Visualization

Finally, we'll run the training loop for a specified number of epochs, calling the `train` and `test` functions in each epoch. We'll also store the loss and accuracy history to plot them afterwards, providing insights into the training process.

````python
epochs = 10
train_losses = []
test_losses = []
test_accuracies = []

for epoch in range(1, epochs + 1):
    train_loss = train(model, device, train_loader, optimizer, epoch)
    test_loss, test_accuracy = test(model, device, test_loader)
    train_losses.append(train_loss)
    test_losses.append(test_loss)
    test_accuracies.append(test_accuracy)

# Plotting the results
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(range(1, epochs + 1), train_losses, label='Training Loss')
plt.plot(range(1, epochs + 1), test_losses, label='Test Loss')
plt.title('Loss over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(range(1, epochs + 1), test_accuracies, label='Test Accuracy', color='green')
plt.title('Test Accuracy over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy (%)')
plt.legend()

plt.tight_layout()
plt.show()
````

### 7. Make Predictions (Optional)

Let's visualize a few predictions from the test set to see how well the model performs on individual examples.

````python
model.eval()

# Get a batch of test data
dataiter = iter(test_loader)
images, labels = next(dataiter)

# Move images to the device for prediction
images = images.to(device)

# Make predictions
outputs = model(images)
_, predicted = torch.max(outputs, 1)

# Move images and predictions back to CPU for plotting
images = images.cpu()
predicted = predicted.cpu()
labels = labels.cpu()

# Plot the images and predictions
fig = plt.figure(figsize=(12, 12))
for i in range(16):
    ax = fig.add_subplot(4, 4, i + 1, xticks=[], yticks=[])
    ax.imshow(images[i].numpy().squeeze(), cmap='gray')
    ax.set_title(f"Pred: {predicted[i].item()} (True: {labels[i].item()})",
                 color=("green" if predicted[i].item() == labels[i].item() else "red"))
plt.tight_layout()
plt.show()
````

# **Question 9: Given a custom image dataset stored in a local directory, write code using Keras ImageDataGenerator to preprocess and train a CNN model.**


### 1. Create a Dummy Dataset (for demonstration purposes)

Since we can't access local file systems directly in this environment, I'll first create a dummy dataset with a directory structure similar to what `ImageDataGenerator` expects. This will involve creating directories for training and validation, and then generating some placeholder image files (empty files in this case, but in a real scenario, these would be actual images).

````python
import os
import numpy as np
from PIL import Image

# Define the base directory for our dummy dataset
data_dir = 'custom_image_dataset'

# Define class names and the number of images per class for train/validation
class_names = ['dog', 'cat', 'bird']
train_images_per_class = 20
val_images_per_class = 5

# Create directories and dummy images
for phase in ['train', 'validation']:
    for class_name in class_names:
        path = os.path.join(data_dir, phase, class_name)
        os.makedirs(path, exist_ok=True)

        num_images = train_images_per_class if phase == 'train' else val_images_per_class
        for i in range(num_images):
            # Create a dummy image (e.g., a small white JPEG image)
            dummy_image = Image.new('RGB', (64, 64), color = 'white')
            image_path = os.path.join(path, f'{class_name}_{i:03d}.jpg')
            dummy_image.save(image_path)

print(f"Dummy dataset created at '{data_dir}' with {len(class_names)} classes.")
````

### 2. Import Libraries

Import the necessary modules from TensorFlow and Keras, along with other utilities.

````python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
import os
````

### 3. Setup ImageDataGenerators for Preprocessing and Augmentation

`ImageDataGenerator` is used to load images from directories, apply real-time data augmentation (for training), and preprocess pixel values. We'll define separate generators for training and validation/testing.

````python
# Define image dimensions and batch size
img_height = 64
img_width = 64
batch_size = 32

# Path to your custom dataset (using the dummy one we created)
train_dir = os.path.join(data_dir, 'train')
validation_dir = os.path.join(data_dir, 'validation')

# Training Data Generator with Augmentation
train_datagen = ImageDataGenerator(
    rescale=1./255,             # Normalize pixel values to [0, 1]
    rotation_range=20,          # Randomly rotate images by up to 20 degrees
    width_shift_range=0.2,      # Randomly shift images horizontally
    height_shift_range=0.2,     # Randomly shift images vertically
    shear_range=0.2,            # Apply shear transformations
    zoom_range=0.2,             # Apply random zoom
    horizontal_flip=True,       # Randomly flip images horizontally
    fill_mode='nearest'         # Strategy for filling in new pixels
)

# Validation Data Generator (only rescaling, no augmentation)
val_datagen = ImageDataGenerator(rescale=1./255)

# Load images from directories using flow_from_directory
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(img_height, img_width),
    batch_size=batch_size,
    class_mode='categorical' # For multi-class classification
)

validation_generator = val_datagen.flow_from_directory(
    validation_dir,
    target_size=(img_height, img_width),
    batch_size=batch_size,
    class_mode='categorical'
)

# Display the class indices found by the generator
print("\nClass indices:", train_generator.class_indices)
````

### 4. Build the CNN Model

We'll define a simple sequential CNN model suitable for classifying images from our custom dataset. The `input_shape` will be `(img_height, img_width, 3)` for RGB images.

````python
model = keras.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(img_height, img_width, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(len(class_names), activation='softmax') # Output layer for number of classes
])

model.summary()
````

### 5. Compile the Model

Compile the model by specifying the optimizer, loss function, and metrics.

````python
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
````

### 6. Train the Model

Train the model using the `fit` method with the `train_generator` and `validation_generator`.

````python
epochs = 10

history = model.fit(
    train_generator,
    epochs=epochs,
    validation_data=validation_generator,
    verbose=1
)

# Plot training history
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.show()
````

### 7. Evaluate the Model

Evaluate the model's performance on the validation set using the `evaluate` method.

````python
loss, accuracy = model.evaluate(validation_generator, verbose=2)
print(f"\nValidation Loss: {loss:.4f}")
print(f"Validation Accuracy: {accuracy:.4f}")
````
